```python
import json
```

```python
topo_india = None
with open("../data/india-udit-001.topo.json", 'r') as f:
    topo_india = json.load(f)
```

```python
print(topo_india.keys())
print(topo_india['objects'].keys())
```

```python
for i in range(len(topo_india['objects']['states']['geometries'])):
    state = topo_india['objects']['states']['geometries'][i]
    state_name = state['properties']['st_nm']
    state_id = state['properties']['st_code']
    del state['properties']['st_nm']
    del state['properties']['st_code']
    state['properties']['id'] = state_id
    state['properties']['name'] = state_name
    state['id'] = state_id
    if state_name == "Lakshadweep" or state_name == "Chandigarh":
        del state['properties']['district']
        del state['properties']['dt_code']
    topo_india['objects']['states']['geometries'][i] = state

for i in range(len(topo_india['objects']['districts']['geometries'])):
    district = topo_india['objects']['districts']['geometries'][i]
    district_name = district['properties']['district']
    
    if district['properties']['st_nm'] == 'Delhi':
        district['properties']['dt_code'] = '91'
    
    district_id = district['properties']['dt_code']
    state_name = district['properties']['st_nm']
    state_id = district['properties']['st_code']
    del district['properties']['district']
    del district['properties']['dt_code']
    del district['properties']['st_nm']
    del district['properties']['st_code']
    district['properties']['id'] = district_id
    district['properties']['name'] = district_name
    district['properties']['state_name'] = state_name
    district['properties']['state_id'] = state_id
    year = district['properties']['year']
    del district['properties']['year']
    district['properties']['year'] = year
    
    # print(district)
    district['id'] = district_id
    topo_india['objects']['districts']['geometries'][i] = district
```
