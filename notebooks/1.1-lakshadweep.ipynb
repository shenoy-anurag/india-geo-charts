```python
import json
import copy
```

```python
topo_lakshadweep = None
with open("../data/lakshadweep.json", 'r') as f:
    topo_lakshadweep = json.load(f)
```

```python
print(topo_lakshadweep.keys())
print(topo_lakshadweep['objects'].keys())
```

```python
topo_lakshadweep['objects']['ind']['geometries'][0]
```

```python
arcs_dweep = topo_lakshadweep['objects']['ind']['geometries'][0]['arcs']
```

```python
chosen_arcs = []
for arc_list in arcs_dweep:
    chosen_arcs.append(topo_lakshadweep['arcs'][arc_list[0][0]])
    print(arc_list[0][0])
```

```python
arcs_dweep
```

```python
topo_lakshadweep['arcs'] = chosen_arcs
j = 0
for i in range(len(arcs_dweep)):
    print(i, arcs_dweep[i][0][0])
    arcs_dweep[i] = [[i]]
```
